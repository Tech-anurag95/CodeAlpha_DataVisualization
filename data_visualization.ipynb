### 1. Environment Setup & Data Loading

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set high-end visual style
sns.set_theme(style='whitegrid', palette='tab10')
plt.rcParams['figure.dpi'] = 150
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

# Load dataset
data_path = '../Order_delivery.csv' if os.path.exists('../Order_delivery.csv') else 'Order_delivery.csv'
df = pd.read_csv(data_path)
df['Order_Time'] = pd.to_datetime(df['Order_Time'], format='%d-%m-%Y %H:%M', errors='coerce')
df['Order_Hour'] = df['Order_Time'].dt.hour
df['Order_Day'] = df['Order_Time'].dt.day_name()
print(f"Loaded {len(df):,} records across {df['City'].nunique()} cities.")
df.head()
```

### 2. Story 1: Geographic Commercial Performance (Revenue vs Transit Times)

```python
city_summary = df.groupby('City').agg({'Total_Price': 'sum', 'Delivery_Duration_Minutes': 'mean'}).reset_index()
city_summary = city_summary.sort_values(by='Total_Price', ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.barplot(data=city_summary, x='Total_Price', y='City', palette='Blues_r', ax=axes[0])
axes[0].set_title('Total Revenue by Market (EGP)', fontweight='bold')

sns.barplot(data=city_summary, x='Delivery_Duration_Minutes', y='City', palette='Reds_r', ax=axes[1])
axes[1].set_title('Average Transit Duration by City (Mins)', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 3. Story 2: Temporal Demand Heatmap (Day of Week × Hour of Day)

```python
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
pivot_tbl = df.pivot_table(index='Order_Day', columns='Order_Hour', values='Order_ID', aggfunc='count').reindex(day_order)

plt.figure(figsize=(13, 6))
sns.heatmap(pivot_tbl, cmap='rocket_r', linewidths=0.5, cbar_kws={'label': 'Orders'})
plt.title('Weekly Order Intensity Heatmap', fontweight='bold', fontsize=13)
plt.xlabel('Hour of Day (0 - 23)')
plt.ylabel('Day of Week')
plt.show()
```

### 4. Story 3: Fleet Vehicle Resilience across Traffic Conditions

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='Driver_Vehicle', y='Delivery_Duration_Minutes', hue='Traffic_Level', palette='coolwarm')
plt.title('Vehicle Transit Times under Traffic Congestion', fontweight='bold')
plt.xlabel('Vehicle Class')
plt.ylabel('Avg Delivery Duration (Minutes)')
plt.legend(title='Traffic Level')
plt.show()
```

### 5. Master C-Suite Visual Analytics Dashboard

```python
# Consolidated 4-Panel Executive Overview
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Panel 1: Top Items
item_vol = df['Item_Name'].value_counts()
sns.barplot(x=item_vol.values, y=item_vol.index, palette='mako', ax=axes[0, 0])
axes[0, 0].set_title('1. Top Revenue-Driving Food Items', fontweight='bold')

# Panel 2: Order Status Share
status_counts = df['Order_Status'].value_counts()
axes[0, 1].pie(status_counts, labels=status_counts.index, autopct='%1.1f%%', colors=['#2ecc71', '#f39c12', '#e74c3c'], startangle=90)
axes[0, 1].set_title('2. Order Completion Funnel', fontweight='bold')

# Panel 3: Payment Preference
sns.countplot(data=df, x='Payment_Method', palette='Set2', ax=axes[1, 0])
axes[1, 0].set_title('3. Customer Payment Channels', fontweight='bold')

# Panel 4: 24-Hour Curve
sns.histplot(df['Order_Hour'], bins=24, kde=True, color='#8e44ad', ax=axes[1, 1])
axes[1, 1].set_title('4. 24-Hour Demand Curve', fontweight='bold')

plt.suptitle('CodeAlpha Data Analytics: Task 3 Executive Dashboard', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
